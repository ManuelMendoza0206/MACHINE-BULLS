## 3. Carga del dataset (solo metadatos JSON)

Usamos `hf_hub_download` en lugar de `datasets.load_dataset()` para evitar el error
"Dataset scripts are no longer supported" y para no descargar los 284 GB.

El JSON de metadatos contiene la estructura de outfits, prendas, categorías y annotations.

*Nombres probables del archivo JSON en el repo:* `mytheresa_outfit_v1.0_2512.json`, `polyvore_outfit_v1.0_2512.json`

```python
from huggingface_hub import hf_hub_download
import json

print("📥 Descargando metadatos de Garments2Look (solo JSON, ~ pocos MB)...")

def cargar_metadata():
    nombres_prueba = [
        "mytheresa_outfit_v1.0_2512.json",
        "polyvore_outfit_v1.0_2512.json",
        "outfits_metadata.json",
        "metadata.json"
    ]

    for filename in nombres_prueba:
        try:
            print(f"🔍 Intentando: {filename}...")
            json_path = hf_hub_download(
                repo_id="ArtmeScienceLab/Garments2Look",
                filename=filename,
                repo_type="dataset"
            )
            with open(json_path, 'r', encoding='utf-8') as f:
                data = json.load(f)
            print(f"✅ Cargado: {filename} ({len(data.get('outfits', []))} outfits)")
            return data
        except Exception as e:
            print(f"❌ {filename} no disponible: {type(e).__name__}")
            continue

    raise RuntimeError("❌ Ningún archivo JSON encontrado. Revisar nombres o subir manualmente.")

data = cargar_metadata()
outfits = data.get('outfits', [])
print(f"\n📊 Total de outfits cargados: {len(outfits)}")
```

## 4. Estadísticas básicas

Ahora vamos a explorar la estructura de los datos cargados.

```python
import numpy as np
from collections import Counter

garments_per_outfit = [len(o.get('garments', [])) for o in outfits]
cnt = Counter(garments_per_outfit)

print(f"👕 Distribución de prendas por outfit: {dict(sorted(cnt.items()))}")
print(f"👕 Promedio: {np.mean(garments_per_outfit):.2f}")
print(f"👕 Mediana: {np.median(garments_per_outfit):.2f}")
print(f"👕 Mín: {min(garments_per_outfit)}, Máx: {max(garments_per_outfit)}")

all_cats = []
for o in outfits:
    for g in o.get('garments', []):
        all_cats.append(g.get('category', 'unknown'))
cat_cnt = Counter(all_cats)
print(f"📂 Top 10 categorías: {cat_cnt.most_common(10)}")
```

## 5. Galería visual de outfits

Mostramos una pequeña muestra de outfits del dataset. Cada outfit
contiene entre 3-12 referencias de prendas (promedio ~4.48).

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.countplot(x=garments_per_outfit, order=sorted(set(garments_per_outfit)))
plt.title('Distribución de número de prendas por outfit')
plt.xlabel('Número de prendas')
plt.ylabel('Cantidad de outfits')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()

plt.figure(figsize=(12, 6))sns.barplot(x=list(cat_cnt.keys())[:10], y=list(cat_cnt.values())[:10])
plt.xticks(rotation=45, ha='right')
plt.title('Top 10 categorías de prenda')
plt.tight_layout()
plt.show()
```

## 6. Subconjunto para MVP (máximo 6 prendas)

Filtramos los outfits que tienen ≤6 prendas, que es el alcance del MVP.

```python
outfits_mvp = [o for o in outfits if len(o.get('garments', [])) <= 6]
mvp_pct = round(100 * len(outfits_mvp) / len(outfits), 1)
print(f"🎯 Outfits con ≤6 prendas (alcance MVP): {len(outfits_mvp)} de {len(outfits)} ({mvp_pct}%)")
```

```python
n = len(outfits)
prom = np.mean(garments_per_outfit)
mvp = len(outfits_mvp)
pct = mvp_pct

print("## Conclusiones del Notebook 01")
print(f"1. Dataset inspeccionado: {n} outfits caracterizados sin descargar las 284 GB.")
print(f"2. Estructura multi-prenda: promedio {prom:.2f} prendas/outfit, rango 3–12.")
print(f"3. Outfits MVP: {pct}% tienen ≤6 prendas, que es el alcance del MVP.")
print(f"4. Dataset no es VON tradicional: La distribución y annotations estructurales justifican por qué StyleSync IA requiere adaptación multi-garment.")
print(f"5. Próximos pasos: Fine-tuning sobre CatVTON → sec. multi-layer (2–3 prendas) → integración con preprocesamiento DWPose → interfaz Next.js.")
```
